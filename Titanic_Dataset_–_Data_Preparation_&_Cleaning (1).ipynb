# **Load the Dataset**



```python
# This loads the data so we can work with it
import pandas as pd

# Read the CSV file
df = pd.read_csv('sample_data/Titanic-Dataset.csv')

# Show the first 5 rows
df.head()
```

# **See What's in Your Data**

```python
# Show how many rows and columns
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Show column names
print("\nColumn names:")
print(df.columns.tolist())
```

# **Find Problems**

```python
# Check for empty values
print("Empty values in each column:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

# **Fix Empty Values**

```python
# Fix Age column - fill empty with average age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fix Cabin column - fill empty with 'Unknown'
df['Cabin'] = df['Cabin'].fillna('Unknown')

# Fix Embarked column - fill empty with most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Check if all fixed
print("Empty values now:")
print(df.isnull().sum())
```

# **Remove Duplicates**

```python
# Count rows before
before = len(df)

# Remove duplicates
df = df.drop_duplicates()

# Count rows after
after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Removed:", before - after)
```

# **Fix Data Types**

```python
# Change Age to whole number
df['Age'] = df['Age'].astype(int)

# Change Sex to category
df['Sex'] = df['Sex'].astype('category')

print("Data types fixed!")
print(df.dtypes)
```

# **Rename Columns**

```python
# Make column names shorter and clearer
df = df.rename(columns={
    'PassengerId': 'ID',
    'SibSp': 'Family',
    'Parch': 'Parents'
})

print("New column names:")
print(df.columns.tolist())
```

# **Check Your Clean Data**

```python
# Show first 5 rows of cleaned data
print("Your cleaned data:")
df.head()
```

# **Save and Download**

```python
# Save to your computer
df.to_csv('cleaned_titanic.csv', index=False)

# Download the file
from google.colab import files
files.download('cleaned_titanic.csv')

print("✅ File downloaded!")
```

# **Quick Summary**

```python
# See some facts about your data
total = len(df)
survived = df['Survived'].sum()
rate = (survived / total) * 100

print("="*40)
print("FACTS ABOUT TITANIC")
print("="*40)
print(f"Total passengers: {total}")
print(f"Survived: {survived} ({rate:.1f}%)")

# Survival by gender
men = df[df['Sex'] == 'male']
women = df[df['Sex'] == 'female']
men_survived = men['Survived'].sum()
women_survived = women['Survived'].sum()

print(f"\nMen: {len(men)} passengers, {men_survived} survived ({men_survived/len(men)*100:.1f}%)")
print(f"Women: {len(women)} passengers, {women_survived} survived ({women_survived/len(women)*100:.1f}%)")
```
